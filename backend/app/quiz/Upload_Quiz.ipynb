# **Upload quiz**: Putting the quiz to MongODB

_Reading in the data..._

```python
# Importing pandas library
import pandas as pd

# Creating the dataframe
df = pd.read_json('questions.json')
df.head(3)
```

_Converting the correct answer..._

```python
# Adding a new column
df['answer'] = df['correct'].apply(lambda x: 0 if x == 'A' else 1
                                   if x == 'B' else 2 if x == 'C' else 'D')

# Checking the answer
df['answer'].value_counts()
```

_Updating the level of hardness..._

```python
# Updating the level
df['level'] = df['level'].apply(lambda x : 0 if x == 'e' else 
                                1 if x == 'm' else 2)

# Checking the level
df['level'].value_counts()
```

_Addressing the answers_

```python
# Identifying the answers
df['A'] = df['options'].apply(lambda x : list(x)[0].replace('A. ', ''))
df['B'] = df['options'].apply(lambda x : list(x)[1].replace('B. ', ''))
df['C'] = df['options'].apply(lambda x : list(x)[2].replace('C. ', ''))
df['D'] = df['options'].apply(lambda x : list(x)[3].replace('D. ', ''))
df.head(3)
```

_Resampling the dataframe..._

```python
# Columns order
col_order = ['level', 'text', 'A', 'B', 'C', 'D', 'answer']

# Resampling the dataframe
df = df[col_order]

# Re-naming text
df.rename(columns = 
    {
        'text' : 'question'
    },
    inplace = True
)
# Checking dataframe
df.head(3)
```

_Connection properties of database..._

```python
# Importing custom pre-process text
import sys
import os
from dotenv import load_dotenv
from pathlib import Path

# Add the 'your_project' root directory to sys.path
# This assumes your_project is the parent of both 'data' and 'utils'
# os.path.abspath(__file__) would give you the notebook's path,
# then two '..' go up to 'your_project'
module_path = os.path.abspath(os.path.join('..', os.path.dirname('')))
if module_path not in sys.path:
    sys.path.append(module_path)

# Importing database connection
from utils.database import Database

# Ensrung we're at the root of the project
project_root = Path(os.getcwd()).resolve().parents[0]

# Constructin the path of the environment variable 
dotenv_path = project_root / 'credentials.env'

# Connecting to the correct environment path
load_dotenv(dotenv_path = dotenv_path)

# Retrieving properties of username
db_password = os.getenv('DB_PASSWORD')
db_string   = os.getenv('DB_STRING')

# Putting the password in the string
db_string = db_string.replace('<db_password>', db_password)

# Name of my database
db_name = 'Hate_App'
    
# Creating the database object
db = Database(connection_string = db_string, db_name = db_name)

# Name of collection to write
col_write = 'questions'
```

_Writing the data..._

```python
# Putting questions in a list
questions = df.to_dict('records')

# Writing the questions
db.add_records(collection_name = col_write, records = questions)
```
