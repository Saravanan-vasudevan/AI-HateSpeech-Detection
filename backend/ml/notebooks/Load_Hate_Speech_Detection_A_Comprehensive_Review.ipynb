## **Paper 1**: Detecting abusive Albanian

```python
# Filename 
fname_daa = 'Detecting_Abusive_Albanian.csv'

# Reading in the dataframe
df_daa = pd.read_csv(fname_daa, sep = ';', header = None, skiprows = 1)

# Addressing column issue
cols_daa = df_daa.columns[:2]

# Resampling the first 2 columns of interest
df_daa  = df_daa[cols_daa]

# Updating the columns
df_daa = df_daa.rename(columns = {
        0 : 'text',
        1 : 'label'
    }
)
# Updating the label column to be a binary classifier
df_daa['label'] = df_daa['label'].apply(lambda x : 0 if x == 'NOT' else 1)

# Adding dataset
df_daa['dataset'] = 'DAA'
df_daa.head(3)
```

## **2**: SWSR

```python
# Filename of the sexism dataset
fname_swsr = 'SWSR.csv'

# Reading in sexism dataset
df_swsr = pd.read_csv(fname_swsr)

# Relevant columns to keep
col_swsr = ['comment_text', 'label']
df_swsr  = df_swsr[col_swsr]

# Updating the text column
df_swsr.rename(
    columns = {
        'comment_text' : 'text'
    },
    inplace = True
)
# Confirming data structue
df_swsr['dataset'] = 'SWSR'
df_swsr.head(3)
```

## **3**: ConvAbuse

```python
# Adding in the convabuse 
fname_convabuse = 'ConvAbuse.csv'

# Reading in the dataframe
df_convabuse = pd.read_csv(fname_convabuse, sep = ';')

# Dropping empty rows
df_convabuse = df_convabuse.dropna(how = 'all')

# Selecting two columns of interest
col_convabuse = ['Input.user', 'is_abuse']
df_convabuse  = df_convabuse[col_convabuse]

# Renaming columns
df_convabuse.rename(
    columns = {
        'Input.user' : 'text',
        'is_abuse'   : 'label'
    },
    inplace = True
)
# Updating the label column
df_convabuse['label'] = df_convabuse['label'].apply(lambda x : 0 if x == 0 else 1)

# Specifying what dataset
df_convabuse['dataset'] = 'ConvAbuse'

# Checking file structure
df_convabuse.head(3)
```

## **5**: SWSR

## **6**: OffensEval

```python
# Name of hugging face dataset
name_offenseval = 'strombergnlp/offenseval_2020'

# Languages
lang_offenseval = ['ar', 'da', 'gr', 'tr']

# List to store outputs
list_offenseval  = []

# Iteraing through languages
for lang in lang_offenseval:

    
    # Retrieving datasets
    ds_splits = load_dataset(name_offenseval, lang)

    # Putting dataset as a list and combining them
    all_splits_list = [ds_splits[split] for split in ds_splits.keys()]
    ds              = concatenate_datasets(all_splits_list)

    # Converting and storing as a list
    list_offenseval.append(ds.to_pandas())

# Creating a dataframe of all languages
df_offenseval = pd.concat(list_offenseval, axis = 0)

# Clearing console
clear_output()

# Selecting right columns
col_offenseval = ['text', 'subtask_a']
df_offenseval  = df_offenseval[col_offenseval]

# Renaming subtask column
df_offenseval.rename(
    columns = {
        'subtask_a' : 'label'
    },
    inplace = True
)
df_offenseval = df_offenseval.dropna()

# Putting label as integer
df_offenseval['label'] = df_offenseval['label'].apply(lambda x : int(x))

# Specifying dataset
df_offenseval['dataset'] = 'OffensEval'
df_offenseval.head(3)
```

## **8**: MultiOff

```python
# Filenames of multioff
fnames_multioff = ['MultiOFF_train.csv', 'MultiOFF_val.csv', 'MultiOFF_test.csv']

# List of multi-off
list_multioff   = []

# Iterating through multi-off
for fname in fnames_multioff:
    list_multioff.append(pd.read_csv(fname))

# Concatenating multi-off
df_multioff = pd.concat(list_multioff, axis = 0)

# Selecting columns of interest
col_multioff = ['sentence', 'label']
df_multioff  = df_multioff[col_multioff]

# Renaming sentence
df_multioff.rename(
    columns = {
        'sentence' : 'text'
    },
    inplace = True
)
# Converting the label
df_multioff['label'] = df_multioff['label'].apply(lambda x : 1 if x == 'offensive' else 0)

# Specfiying what dataset
df_multioff['dataset'] = 'MultiOFF'

# Checking top of 3 rows
df_multioff.head(3)
```

## **9**: Ethos

```python
# Filename of ethos
fname_ethos = 'Ethos.csv'

# Reading in dataframe
df_ethos = pd.read_csv(fname_ethos, sep = ';')

# Renaming columns
df_ethos.rename(
    columns = {
        'comment' : 'text',
        'isHate'  : 'label'
    },
    inplace = True
)
# Converting label
df_ethos['label'] = df_ethos['label'].astype('int')

# Specifying which dataset
df_ethos['dataset'] = 'Ethos'

# Checking top of dataframe
df_ethos.head()
```

## **10**: LHSAB

```python
# Reading in lsab
fname_lhsab = 'LHSAB.txt'

# Reading in the dataframe
df_lhsab = pd.read_csv(fname_lhsab, sep = '\t')

# Renaming columns
df_lhsab.rename(
    columns = {
        'Tweet' : 'text',
        'Class' : 'label'
    },

    inplace = True
)
# Specifying label
df_lhsab['label'] = df_lhsab['label'].apply(lambda x : 1 if x == 'abusive' else 0)

# Adding dataset
df_lhsab['dataset'] = 'LH-SAB'

# Checking structure
df_lhsab.head(3)
```

## **11**: MLD

```python
# Filename of mld
fname_mld = 'MLD.csv'

# Reading in MLD dataset
df_mld = pd.read_csv(fname_mld, encoding_errors = 'replace')

# Selecting two columns
col_mld = ['Tweet', 'HS']
df_mld  = df_mld[col_mld]

# Renaming columns
df_mld.rename(
    columns = {
        'Tweet' : 'text',
        'HS'    : 'label'
    },
    inplace = True
)
# Specifying what dataset
df_mld['dataset'] = 'MLD'

# Checking top of dataframe
df_mld.head(3)
```

## **12**: ALD

```python
# Reading in ALD
fname_ald = 'ALD.csv'

# Reading in dataframe
df_ald = pd.read_csv(fname_ald)

# Selecting columns of interest
col_ald = ['Tweet', 'Label']
df_ald  = df_ald[col_ald]

# Converting columns
df_ald.rename(
    columns = {
        'Tweet' : 'text',
        'Label' : 'label'
    },
    inplace = True
)
# Converting the label
df_ald['label'] = df_ald['label'].apply(lambda x : 1 if x == 2 else 0)

 # Specifying dataset
df_ald['dataset'] = 'ALD'

# Checking top of dataframe
df_ald.head()
```

# **Saving data**

_Storing all dataframes..._

```python
# Concatenating dataframes
df = pd.concat(
    [df_ald, df_daa, df_ethos, df_lhsab,
     df_mld, df_multioff, df_offenseval, 
     df_swsr, df_convabuse, df_dynabench
    ],
    axis = 0   
)
```

## **Process data**

* I am going to perform some mild modification to my dataframe. This includes
> * Address emojis and obfuscation
> * Detect what language the tweets are

```python
# Dropping null observations
df = df.dropna()

# Introduced the new processed tweets
df['processed_text'] = df['text'].apply(lambda x : preprocess_for_hate_speech(text = x))

# Checking what language
# Here, we are using full name node the code
df['language'] = df['text'].apply(lambda x : detect_language(text = x))
```
